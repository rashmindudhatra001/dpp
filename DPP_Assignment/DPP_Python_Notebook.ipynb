## 3.a. Missing value Imputation

```python
#before missing value imputation
missing_values = df.isnull().sum()
print("before missing value imputation:\n",missing_values[missing_values >0])

#1. Fill missing values in numerical columns using the median
for column in numerical_columns:
    df_clean[column].fillna(df_clean[column].median(),inplace =True)

#2. Fill missing values in categorical columns using the mode
for column in categorical_columns:
  df_clean[column].fillna(df_clean[column].mode()[0],inplace=True)

#after missing vallue imputation
print("\n after missing value imputation:\n",df_clean.isnull().sum())
```

Missing Value Imputation: Missing values were handled by replacing numerical missing values with the median and categorical missing values with the mode. The median was chosen because it is less affected by outliers than the mean, while the mode represents the most frequent category.After doing missing value imputation , there is zero missing values in each row

```python
#After Standardization

#Remove leading and trailing spaces
for column in columns_to_standardize:
    df_clean[column] = df_clean[column].str.strip()

#Standardise Gender Column
df_clean["Gender"] = df_clean["Gender"].replace({
    "MALE": "Male",
    "male": "Male",
    "M": "Male",
    "FEMALE": "Female",
    "female": "Female",
    "F": "Female"
})


#Standardise Income Currency Column
df_clean["IncomeCurrency"] = df_clean["IncomeCurrency"].replace({
    "₹": "INR",
    "Rs": "INR"
})


#Standardise Membership Tier
df_clean["MembershipTier"] = df_clean["MembershipTier"].replace({
    "PLATINUM": "Platinum",
    "basic": "Basic",
    "silver": "Silver",
    "Silvr": "Silver",
    "GOLD": "Gold"
})

#Standardise Visit Frquency
df_clean["VisitFrequency"] = df_clean["VisitFrequency"].str.title()


#Standardise Preffered Category
df_clean["PreferredCategory"] = df_clean["PreferredCategory"].str.title()

df_clean["PreferredCategory"] = df_clean["PreferredCategory"].replace({
    "Grocerry": "Grocery",
    "Groceries": "Grocery",
    "Electronic": "Electronics",
    "Eletronics": "Electronics"
})


#Standardise City Category
df_clean["City"] = df_clean["City"].str.title()
df_clean["City"] = df_clean["City"].replace({
    "Chenai": "Chennai",
    "Calcutta": "Kolkata",
    "Bangalore": "Bengaluru",
    "Mumbay": "Mumbai",
    "New Delhi": "Delhi",
    "Hyd": "Hyderabad",
    "Hyderbad": "Hyderabad"
})

#Standardise Country Category
df_clean["Country"] = df_clean["Country"].replace({
    "INDIA": "India",
    "india": "India",
    "IN": "India",
    "Bharat": "India"
})

#Standardise EmailProvider Category
df_clean["EmailProvider"] = df_clean["EmailProvider"].str.lower()

#Standardise PaymentMethod Category
df_clean["PaymentMethod"] = df_clean["PaymentMethod"].replace({
    "upi": "UPI",
    "CREDIT CARD": "Credit Card"
})

#Standardise Couponused Category
df_clean["CouponUsed"] = df_clean["CouponUsed"].replace({
    "yes": "Yes",
    "Y": "Yes",
    "1": "Yes",
    "no": "No",
    "N": "No",
    "0": "No"
})

# Standardize DeviceType values
df_clean["DeviceType"] = (
    df_clean["DeviceType"]
    .str.strip()           # Remove leading/trailing spaces
    .replace({
        "IOS": "iOS",
        "android": "Android",
        "web": "Web",
    })
)


#final output after standardisation with frequency
print("\n Dispaying the frequency of unique values in each column after standardisation: \n")
for column in columns_to_standardize:
    print("=" * 50)
    print(f"Column: {column}")
    print(df_clean[column].value_counts(dropna=False))
```

Reason: Standardization was performed to convert inconsistent representations of the same category into a single, consistent format. This improves data quality, eliminates duplicate categories caused by differences in spelling, capitalization, or spacing, and ensures accurate analysis and machine learning.

## 3.f. Outlier Treatment


```python
#Before -visualise them, Count the outliers before treatment
import matplotlib.pyplot as plt

numeric_columns = [
    "Age",
    "AnnualIncome_INR",
    "AvgBasketValue_INR",
    "LoyaltyPoints"
]

for column in numeric_columns:
    plt.figure(figsize=(5,3))
    plt.boxplot(df_clean[column], vert=False)
    plt.title(f"Box Plot of {column}")
    plt.xlabel(column)
    plt.show()
```

```python
#Outlier Treatment using IQR method
outlier_columns = [
    "Age",
    "AnnualIncome_INR",
    "AvgBasketValue_INR",
    "LoyaltyPoints"
]

for column in outlier_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean[column] = df_clean[column].clip(lower, upper)
```

```python
#After
for column in outlier_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower) |
        (df_clean[column] > upper)
    ]

    print(column, ":", outliers.shape[0])
```

Reason:Outliers were treated using the Interquartile Range (IQR) method. Instead of removing records, extreme values were capped at the lower and upper IQR limits using the clip() function. This reduces the influence of outliers while preserving all observations in the dataset.
After outlier treatment, the selected numerical columns no longer contained values outside the IQR limits, resulting in a cleaner and more reliable dataset for further analysis.

### Selection of Data Transformation Techniques
To prepare the cleaned dataset for machine learning, three data transformation techniques were selected: Label Encoding, One-Hot Encoding, and Min-Max Normalization. Label Encoding was used for binary categorical variables, while One-Hot Encoding was applied to categorical variables with multiple categories to convert them into numerical features without introducing an artificial order. Min-Max Normalization was chosen to scale numerical features to a common range (0–1), ensuring that variables with larger values do not dominate the learning process. These transformations improve data consistency and make the dataset suitable for machine learning algorithms.

# Task 5 Data Reduction
Data reduction was performed using PCA and Random Sampling to reduce the dataset size while preserving most of the important information.

## 5.a.PCA
PCA was applied to the numerical features to reduce dimensionality while preserving most of the information.

```python
#Before
# Select numerical columns for PCA
pca_columns = [
    "Age",
    "AnnualIncome_INR",
    "SpendingScore_1_100",
    "AvgBasketValue_INR",
    "TotalPurchases",
    "OnlinePurchases",
    "StorePurchases",
    "LoyaltyPoints"
]

print("Number of features before PCA:", len(pca_columns))
display(df_clean[pca_columns].head())
```

```python
#after
from sklearn.decomposition import PCA

# Create PCA object
pca = PCA(n_components=5)

# Apply PCA
pca_result = pca.fit_transform(df_clean[pca_columns])

# Create a new DataFrame
df_pca = pd.DataFrame(
    pca_result,
    columns=["PC1", "PC2", "PC3", "PC4", "PC5"]
)

display(df_pca.head())

print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Explained Variance:")
print(pca.explained_variance_ratio_.sum())
```

```python
pca_loadings = pd.DataFrame(
    pca.components_.T,
    index=pca_columns,
    columns=["PC1","PC2","PC3","PC4","PC5"]
)

display(pca_loadings.round(3))
```

# Task 6 Proximity Measures

## 6.a. Euclidean Distance
Euclidean distance measures the straight-line distance between two data points in a multi-dimensional space. It is one of the most commonly used proximity measures, where smaller distances indicate that the data points are more similar.

```python
#Before
print("Original dataset shape:", df_clean.shape)

# Select 20 random customers
df_20 = df_clean.sample(
    n=20,
    random_state=42
).reset_index(drop=True)

print("Selected dataset shape:", df_20.shape)
display(df_20.head())
```

```python
#Select numerical columns for proximity calculation

proximity_columns = [
    "Age",
    "AnnualIncome_INR",
    "SpendingScore_1_100",
    "AvgBasketValue_INR",
    "TotalPurchases",
    "OnlinePurchases",
    "StorePurchases",
    "LoyaltyPoints"
]

X = df_20[proximity_columns]
display(X.head())

from sklearn.metrics import pairwise_distances

# Calculate Euclidean distance
euclidean_matrix = pairwise_distances(
    X,
    metric="euclidean"
)

#convert to dataframe
euclidean_df = pd.DataFrame(
    euclidean_matrix,
    index=df_20.index,
    columns=df_20.index
)

display(euclidean_df.round(2))
```

The Euclidean distance matrix was calculated for the selected 20 customers using their normalized numerical features. Smaller distance values indicate greater similarity between customers, while larger values indicate greater dissimilarity.
Customers 10 and 18 are the most similar customers based on the Euclidean distance, with the smallest distance of 0.21. This indicates that these two customers have the most similar normalized numerical feature values among the selected 20 customers.

## 6.b.Manhattan Distance
Manhattan distance measures the distance between two data points by summing the absolute differences of their feature values. It represents movement along horizontal and vertical paths rather than a straight line, with smaller values indicating greater similarity.

```python
from sklearn.metrics import pairwise_distances

# Calculate Manhattan distance
manhattan_matrix = pairwise_distances(
    X,
    metric="manhattan"
)

# Convert to DataFrame
manhattan_df = pd.DataFrame(
    manhattan_matrix,
    index=df_20.index,
    columns=df_20.index
)

display(manhattan_df.round(2))
```

The Manhattan distance matrix was analyzed to identify similar customers. The smallest non-zero distance was 0.49 between Customers 10 and 18, indicating that these two customers are the most similar based on the selected normalized numerical features.Both Euclidean and Manhattan distance measures identified Customers 10 and 18 as the most similar pair, indicating consistent similarity across different proximity measures.
